```python
from dotenv import load_dotenv
import os

load_dotenv("/Users/cheranreddy/PycharmProjects/Learning_Agentic_AI/.env")

print("Tracing:", os.getenv("LANGCHAIN_TRACING_V2"))
print("Project:", os.getenv("LANGCHAIN_PROJECT"))
print("API key:", os.getenv("LANGCHAIN_API_KEY")[-7:])
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

agent= create_agent(model='gpt-5-mini')
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]}
    # No config= here — stateless agents don't use thread_id
)
```

```python
from pprint import pprint
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]}  # Only this one message — no history!
)

pprint(response)
```

```python
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent
from langchain.messages import HumanMessage

agent= create_agent(model='gpt-5-mini',
                   checkpointer=InMemorySaver())

```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

config = {"configurable":{"thread_id":1}}

response = agent.invoke(
    {"messages":[question]},
     config,
)
```

```python
question = "what is my favourite colour"
config = {"configurable":{"thread_id":1}}
response = agent.invoke(
    {"messages":[question]},
    config
)
```

```python
from pprint import pprint
pprint(response['messages'][-1].content)
```
